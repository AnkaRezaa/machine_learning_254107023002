```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Load dataset
df = pd.read_csv('spam.csv', encoding='latin-1')

# Cleaning & Preprocessing awal
df = df.drop(df.iloc[:, 2:], axis=1)
df.columns = ['Labels', 'SMS']

# Encoding label (ham = 0, spam = 1)
df['Labels'] = df['Labels'].map({'ham': 0, 'spam': 1})

# Split dataset (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(
    df['SMS'], df['Labels'], test_size=0.2, random_state=42, stratify=df['Labels']
)
```

```python
# Ekstraksi fitur TfidfVectorizer dengan mengaktifkan stop_words
tfidf_vec = TfidfVectorizer(stop_words='english')
X_train_tfidf = tfidf_vec.fit_transform(X_train)
X_test_tfidf = tfidf_vec.transform(X_test)

# Pelatihan model Multinomial Naive Bayes
mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train)

# Prediksi & Evaluasi
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)

print("=== EVALUASI MODEL 2 (TF-IDF + Stop Words) ===")
print(f"Akurasi: {acc_tfidf * 100:.2f}%\n")
print(classification_report(y_test, y_pred_tfidf, target_names=['Ham', 'Spam']))
```

### Kesimpulan Analisis:

1. **Perbandingan Hasil**:
   * **CountVectorizer** menghasilkan nilai **Akurasi** dan **Recall (Spam)** yang lebih tinggi dibanding TF-IDF.
   * **TF-IDF** memiliki **Precision** yang sempurna (100%), yang berarti tidak ada pesan `ham` yang salah ditebak sebagai `spam`. Namun, *Recall*-nya lebih rendah karena TF-IDF menekan bobot kata-kata yang sering muncul di banyak dokumen.

2. **Fitur Terbaik untuk Dataset `spam.csv`**:
   * **CountVectorizer (dengan stop_words)** adalah fitur **terbaik** untuk kasus klasifikasi spam pada dataset ini.
   * **Alasan**: Pada pendeteksian spam, *Recall* sangat krusial agar kata-kata indikator spam (seperti *free*, *claim*, *call*, *txt*) tidak berbobot terlalu kecil. Frekuensi kemunculan kata secara langsung (*word count*) pada Multinomial Naive Bayes memberikan pembobotan probabilitas yang lebih efektif dibanding pembobotan frekuensi terbalik (*inverse document frequency*) dari TF-IDF.
