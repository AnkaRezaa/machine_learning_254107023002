3. Kegunaan

- PyPREP (pyprep)

Kegunaan: Library Python berbasis MNE-Python yang digunakan untuk preprocessing data sinyal EEG (Elektroensefalogram).

Fungsi Utama: Melakukan pembersihan data EEG secara otomatis, seperti deteksi dan penghapusan channel yang bermasalah (bad channels), interpolasi sinyal, serta penerapan referensi ulang Averaged Reference (PREP pipeline).

- SciPy (scipy)

Kegunaan: Ecosystem library fundamental untuk komputasi ilmiah (scientific computing) dan matematika tingkat tinggi berbasis NumPy.

Fungsi Utama: Menyediakan fungsi-fungsi pemrosesan sinyal (scipy.signal), aljabar linier (scipy.linalg), optimasi, integrasi numerik, dan analisis statistik yang sering digunakan dalam pra-pengolahan data data science / machine learning.

- Weights & Biases (wandb)

Kegunaan: Platform MLOps (Machine Learning Operations) untuk pelacakan eksperimen (experiment tracking) dan visualisasi model machine learning.

Fungsi Utama: Mencatat hyperparameter, metric performa model (seperti loss dan accuracy), menyimpan checkpoint dataset/model (artifact), serta membandingkan performa berbagai eksperimen model secara visual melalui dashboard.

- pyECG (pyECG)

Kegunaan: Library Python khusus yang dirancang untuk analisis dan pemrosesan sinyal ECG/EKG (Elektrokardiogram atau rekam jantung).

Fungsi Utama: Mengekstrak fitur-fitur penting dari sinyal EKG, mendeteksi puncak gelombang (QRS complex / R-peak detection), menghitung variabilitas detak jantung (Heart Rate Variability / HRV), dan membersihkan noise pada sinyal EKG.
